# 03 - Baseline Regression Models

Compare fast baseline recipes using the shared pipeline.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

import pandas as pd

from src.data import load_raw_data, split_features_target
from src.evaluate import cross_validate_pipeline
from src.models import build_model_pipeline

train, test, _ = load_raw_data()
X, y = split_features_target(train)
```

```python
results = []
for model_name in ['ridge', 'hgb']:
    pipeline = build_model_pipeline(X, model_name=model_name, random_state=42)
    _, fold_metrics, summary = cross_validate_pipeline(pipeline, X, y, n_splits=5, random_state=42)
    summary['model_name'] = model_name
    results.append(summary)

pd.DataFrame(results).sort_values('rmse')
```
