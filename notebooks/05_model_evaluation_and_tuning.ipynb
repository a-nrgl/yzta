# 05 - Model Evaluation and Tuning

Review fold metrics, residuals, and prediction behavior from the latest training run.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.append(str(PROJECT_ROOT))

import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

metrics = json.loads((PROJECT_ROOT / 'reports/metrics.json').read_text())
fold_metrics = pd.read_csv(PROJECT_ROOT / 'reports/fold_metrics.csv')
oof = pd.read_csv(PROJECT_ROOT / 'reports/oof_predictions.csv')
metrics, fold_metrics
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(oof['residual'], bins=50, kde=True, ax=axes[0])
axes[0].set_title('OOF residuals')
sns.scatterplot(data=oof.sample(min(5000, len(oof)), random_state=42), x='prediction', y='residual', alpha=0.35, ax=axes[1])
axes[1].axhline(0, color='black', linewidth=1)
axes[1].set_title('Residuals vs prediction')
plt.tight_layout()
plt.show()
```
